Soal 1 : Belanja Buku

Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

Tentukan:

1. Total belanja, besar diskon, dan jumlah yang harus dibayar.
2. Uang kembalian.
3. Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat, berapa iuran per orang dan berapa rupiah sisanya?
4. Apakah uang Budi cukup? (hasil True/False)

```python
# ============================================================
# Jawaban Soal 1 - Belanja Buku
# ============================================================

# Nama   : Intan Dwinancy Aprilia Nainggolan
# NIM    : 4232501035
# Tujuan : Menghitung total belanja, diskon, uang kembalian,
#          iuran per orang, sisa pembagian, dan kecukupan uang.

# ============================================================

# Data belanja
jumlah_buku = 7
harga_per_buku = 18500
diskon_persen = 10
uang_budi = 150000
jumlah_orang = 4

# 1. Menghitung total belanja, diskon, dan pembayaran
total_belanja = jumlah_buku * harga_per_buku
diskon = total_belanja * diskon_persen // 100
total_bayar = total_belanja - diskon

# 2. Menghitung uang kembalian
kembalian = uang_budi - total_bayar

# 3. Menghitung iuran per orang dan sisa pembagian
iuran_per_orang = total_bayar // jumlah_orang
sisa = total_bayar % jumlah_orang

# 4. Memeriksa apakah uang Budi cukup
uang_cukup = uang_budi >= total_bayar

# Menampilkan hasil
print("=== BELANJA BUKU ===")
print("Total belanja       : Rp", total_belanja)
print("Besar diskon        : Rp", diskon)
print("Total yang dibayar  : Rp", total_bayar)
print("Uang kembalian      : Rp", kembalian)
print("Iuran per orang     : Rp", iuran_per_orang)
print("Sisa pembagian      : Rp", sisa)
print("Uang Budi cukup?    :", uang_cukup)
```

Soal 2 : PLTS Atap Rumah

Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8). Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.

Tentukan:

1. Energi yang dihasilkan per hari (kWh).
2. Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan.
3. Kekurangan energi per hari terhadap kebutuhan.
4. Apakah PLTS sudah mencukupi kebutuhan rumah? (True/False)

```python
# ============================================================
# Jawaban Soal 2 - PLTS Atap Rumah
# ============================================================

# Nama   : Intan Dwinancy Aprilia Nainggolan
# NIM    : 4232501035
# Tujuan : Menghitung energi listrik harian dan bulanan PLTS,
#          nilai rupiah listrik yang dihasilkan, kekurangan
#          energi, dan kecukupan kebutuhan listrik rumah.

# ============================================================

# Data PLTS dan Kebutuhan Listrik
jumlah_panel = 12
daya_panel = 450
penyinaran = 5
efisiensi = 0.8
kebutuhan_harian = 25
tarif_listrik = 1444.70

# 1. Menghitung energi yang dihasilkan per hari (kWh)
daya_total = jumlah_panel * daya_panel
energi_harian = daya_total * penyinaran * efisiensi / 1000

# 2. Menghitung energi per bulan dan nilai rupiah
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

# 3. Menghitung kekurangan energi per hari
kekurangan_energi = kebutuhan_harian - energi_harian

# 4. Memeriksa kecukupan kebutuhan listrik rumah
plts_cukup = energi_harian >= kebutuhan_harian

# Menampilkan Hasil
print("=== HASIL PERHITUNGAN PLTS ATAP RUMAH ===")
print(f"Energi per hari       : {energi_harian:.2f} kWh")
print(f"Energi per bulan      : {energi_bulanan:.2f} kWh")
print(f"Nilai listrik per bulan: Rp{nilai_listrik:,.2f}")
print(f"Kekurangan energi     : {kekurangan_energi:.2f} kWh/hari")
print(f"PLTS mencukupi?       : {plts_cukup}")
```

Soal 3 : Generator Diesel dan Rugi Kabel

Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

Tentukan:

1. Energi yang dibangkitkan per hari (kWh).
2. Kebutuhan solar per hari dan biayanya.
3. Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
4. Rugi daya pada kabel (P = I² × R).
5. Apakah rugi kabel tergolong tinggi, yaitu lebih dari 150 W dan arus minimal 20 A? (True/False)

```python
# ============================================================
# Jawaban Soal 3 - Generator Diesel dan Rugi Kabel
# ============================================================

# Nama   : Intan Dwinancy Aprilia Nainggolan
# NIM    : 4232501035
# Tujuan : Menghitung energi harian generator, kebutuhan dan
#          biaya solar, kecukupan tangki, rugi daya kabel,
#          serta menentukan apakah rugi kabel tergolong tinggi.

# ============================================================

# Data Generator Diesel
daya_generator = 200
jam_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000

# Data Kabel
arus = 20
hambatan = 0.5

# 1. Menghitung energi yang dibangkitkan per hari
energi_harian = daya_generator * jam_operasi

# 2. Menghitung kebutuhan solar per hari dan biayanya
solar_harian = energi_harian * konsumsi_solar
biaya_solar = solar_harian * harga_solar

# 3. Menghitung jumlah hari penuh tangki cukup dan sisa solar
hari_penuh = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

# 4. Menghitung rugi daya pada kabel
rugi_daya = arus ** 2 * hambatan

# 5. Memeriksa apakah rugi kabel tergolong tinggi
rugi_tinggi = rugi_daya > 150 and arus >= 20

# Menampilkan Hasil
print("=== HASIL PERHITUNGAN GENERATOR DIESEL DAN RUGI KABEL ===")
print(f"Energi per hari       : {energi_harian} kWh")
print(f"Kebutuhan solar       : {solar_harian:.0f} liter/hari")
print(f"Biaya solar per hari  : Rp{biaya_solar:,.0f}")
print(f"Tangki cukup          : {hari_penuh} hari penuh")
print(f"Sisa solar            : {sisa_solar:.0f} liter")
print(f"Rugi daya kabel       : {rugi_daya:.0f} W")
print(f"Rugi kabel tergolong tinggi? : {rugi_tinggi}")
```
